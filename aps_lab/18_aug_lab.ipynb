```python
import matplotlib.pyplot as plt
```

```python
num_people=list(range(1,81))
prob_no_match=1.0
probabilities=[]
```

```python
for i in num_people:
    prob_no_match*=(365-i+1)/365
    prob_match=1.0-prob_no_match
    probabilities.append(prob_match)
```

```python
plt.figure(figsize=(10,6))
plt.plot(num_people,probabilities,color='blue',linewidth=2.5,label='probability of match')
```
