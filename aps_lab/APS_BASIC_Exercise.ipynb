    Addition rule for disjoint events

    A die is rolled 10,000 times.

    Define: A=getting 1, B= getting 6, check the events are joint or disjoint  

    Verify: P(A∪B)=P(A)+P(B)

```python
prob_getting_1=rd.randint(1,10000)
prob_getting_2=rd.randint(1,10000)
#Events are disjoint because A={1} , b={6} therefore A ∩ B = Null
P_of_A = prob_getting_1/10000
P_of_B = prob_getting_2/10000
PA_plus_PB = P_of_A + P_of_B
print(f"P(A∪B)=P(A)+P(B) is : {PA_plus_PB}")
```

    Addition rule for overlapping events

```python
# A=getting an even number A={2,4,6}
# B=getting a number greater than 3 B={4,5,6}
p_of_A = 3
p_of_B = 3
p_of_A_internsection_B = 2 #{4,6}
cal=p_of_A + p_of_B - p_of_A_internsection_B
print(f"P(A∪B)=P(A)+P(B)−P(A∩B): P(A∪B)={cal}")
```
