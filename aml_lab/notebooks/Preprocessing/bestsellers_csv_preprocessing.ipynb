```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.preprocessing import StandardScaler
ct = ColumnTransformer(
    transformers=[
        ("author_enc", OneHotEncoder(handle_unknown='ignore'), [0]), # Author
        ("year_enc", OneHotEncoder(), [4]),                         # Year
        ("scaler", StandardScaler(), [1, 2, 3])                     # Rating, Reviews, Price
    ],
    remainder="drop" # or "passthrough" if you keep Name
)

X_transformed = ct.fit_transform(X)
```

```python
pd.DataFrame(X_transformed).head()
```

```python
import pandas as pd

# 1. Get the new column names generated by the ColumnTransformer
feature_names = ct.get_feature_names_out()

# 2. Convert the sparse matrix to a dense array and load it into a DataFrame
df_encoded = pd.DataFrame(X_transformed.toarray(), columns=feature_names)

# 3. View the first few rows
print(df_encoded.head())
```
