```python
data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya",
        "John", "Priya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE",
        "ME", "ECE"
    ],

    "Attendance": [
        92, 75, 96, 68,
        82, 88, 72, 95
    ],

    "Internal": [
        85, 72, 94, 65,
        80, 86, 70, 91
    ],

    "FinalMarks": [
        88, 74, 96, 62,
        83, 89, 68, 93
    ]
}

df = pd.DataFrame(data)
df
```

```python
df.head()
```

```python
df.shape
```

```python
df["FinalMarks"].mean()
```

```python
df.loc[df["FinalMarks"].idxmax()]
```

```python
df.loc[df["FinalMarks"].idxmin()]
```

```python
df[df["FinalMarks"]>80]
```

```python
df.groupby("Department")["FinalMarks"].mean()
```

```python
df.sort_values("FinalMarks")
```

```python
df["Result"]=np.where(df["FinalMarks"]>=70,"Pass","Fail")
df
```

```python
conditions=[df["FinalMarks"]>=90,df["FinalMarks"]>=80,df["FinalMarks"]>=70]
categories=["Excellent","Very Good","Good"]

df["Performance"]=np.select(conditions,categories,default="Needs Improvement")
df
```

```python
avg_by_dept=df.groupby("Department")["FinalMarks"].mean()
avg_by_dept
```
