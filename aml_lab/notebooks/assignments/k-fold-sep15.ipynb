```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score, StratifiedKFold
```

```python
df=pd.read_csv("/home/msis/aml_lab/data/Social_Network_Ads.csv")
df
```

```python
X=df.iloc[:,2:4]
X
```

```python
y=df.loc[:,['Purchased']]
y
```

```python
dt_classifier = DecisionTreeClassifier(random_state=42, max_depth=3)

kf = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
```

```python
cv_scores = cross_val_score(dt_classifier, X, y, cv=kf, scoring='accuracy')

print("Accuracy scores for each fold:", cv_scores)
print(f"Average Accuracy: {cv_scores.mean() * 100:.2f}%")
```
