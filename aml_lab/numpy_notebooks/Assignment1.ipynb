```python
#2. Write code using lambda function that takes in three parameters and adds the first two.
a=1
b=2
c=3
result=(lambda x,y,z:x+y)(a,b,c)
print(result)
```

```python
#4. Program to double each item in a list using map()
nums=[1,2,3,4]
result=map(lambda x:x*x,nums)
print(result)

for i in result:
    print(i)

print(list(result))
```
