8. Given:
salary = np.array([25000, 32000, 28000, 45000,
                   50000, 38000, 42000, 30000])

Find:

Total salary
Average salary
Highest salary
Lowest salary
Median salary
Standard deviation

```python
salary = np.array([25000, 32000, 28000, 45000,
                   50000, 38000, 42000, 30000])
print("Total salary: ",salary.sum())
print("Average salary: ",salary.sum()/len(salary))
print("Highest salary: ",salary.max())
print("Lowest salary: ",salary.min())
print("Median salary ",np.median(salary))
print("Standard deviation",salary.std())
```

9. Given: marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])

Using NumPy boolean indexing, find:

Marks greater than 75
Marks between 50 and 80
Number of students who scored below 50
Percentage of students who scored 60 or above

Do not use a Python for loop.

```python
marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])
print("Marks greater than 75 :",marks[marks>75])
print("Marks between 50 and 80 :",marks[(marks>50) & (marks>80)])
print("Number of students who scored below 50 :",marks[marks<50])
print("Marks greater than 60 :",marks[marks>=60])
print("Percentage of students who scored 60 or above:", len(marks[marks>=60])/len(marks)*100)
```
