```python
#An example of a class in python
class Person():
  institute = 'MSIS' # class variable
  #self keyword to access the attributes and methods of the class
  def set_name(self, new_name): # method
      self.name = new_name
  def set_location(self, new_location):
      self.location=new_location
```

```python
detail = Person()
detail.set_name('Kushal')
detail.set_location('Manipal')
```

```python
print('{} lives in {} and studies in the institute {}'.format(detail.name, detail.location, detail.institute))
```

```python
#mapping the min function between three lists
store1=[10.00, 12.45, 6.04, 5.55]
store2=[22.33, 12.22, 5.77, 3.44]
store3=[13.22, 12.45, 6.57, 4.55]
cheapest = map(min, store1, store2, store3)
cheapest = map(max, store1, store2, store3)
```

```python
print(cheapest)
```

```python
for item in cheapest:
  print(item)
```

```python
def calculateSquare(n):
  return n*n
```

```python
numbers = (1,2,3,4)
result=map(calculateSquare, numbers)
print(result)
```

```python
for i in result:
  print(i)
```

```python
res=list(result)
print(res)
```

```python
numbers = (1,2,3,4)

result = map(lambda x:x*x,numbers) #lambda is a shortlived function
print(result)
```
